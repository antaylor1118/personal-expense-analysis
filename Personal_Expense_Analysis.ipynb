## Data Preparation
This section creates a sample expense dataset
and organizes it into a Pandas DataFrame.

```python
expenses = [250, 120, 85, 70, 50, 95, 60]
print(expenses)
```

```python
total_expenses=sum(expenses)
print(total_expenses)
```

```python
average_expenses=(sum(expenses)/len(expenses))
print(average_expenses)
```

```python
highest_expense=max(expenses)
print(highest_expense)

lowest_expense=min(expenses)
print(lowest_expense)
```

```python
import pandas as pd
```

```python
data={
    'Category':['Groceries','Gas','Shopping','Internet','Entertainment','Restaurant','Transportation'],
    'Amount':[250, 120, 85, 70, 50, 95, 60]
    }
print(data)
```

```python
df=pd.DataFrame(data)
display(df)
```

## Data Analysis
This section calculates the total, average,
highest, and lowest expenses.

```python
total=df['Amount'].sum()
print(total)
```

```python
average=df['Amount'].mean()
print(average)
```

```python
highest=df['Amount'].max()
print(highest)

Lowest=df['Amount'].min()
print(lowest)
```

```python
highest_index=  df['Amount'].idxmax()
print(highest_index)

lowest_index=df['Amount'].idxmin()
print(lowest_index)
```

```python
highest_category = df.loc[highest_index, 'Category']
print(highest_category)

lowest_category = df.loc[lowest_index, 'Category']
print(lowest_category)
```

## Data Visualization
This section creates a bar chart to compare
expenses across different categories.

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(df['Category'], df['Amount'])
plt.show()
```

```python
plt.bar(df['Category'], df['Amount'])
plt.title('Monthly Expenses by Category')
plt.xlabel('Expense Category')
plt.ylabel('Amount($)')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.bar(df['Category'], df['Amount'])
plt.title('Monthly Expenses by Category')
plt.xlabel('Expense Category')
plt.ylabel('Amount($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.bar(df['Category'], df['Amount'])
plt.title('Monthly Expenses by Category')
plt.xlabel('Expense Category')
plt.ylabel('Amount($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('monthly_expenses.png',dpi=300)
plt.show()
```

## Key Findings
The total expenses across all seven categories are $730. Groceries has the highest spending at $250, while Entertainment has the lowest spending at $50. The bar chart makes it easier to compare spending between different categories. I also learned how to create, organize, and save data charts using Python.
